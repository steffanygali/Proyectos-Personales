```python
#PROGRAMA 1 DE STEFF
import pandas as pd
est = pd.read_csv('data_studentlist.csv')
print('datos listos y preparados si que si:D')
est.head(5)
hom = est[est['Gender'] == 'M']
muj = est[est['Gender'] == 'F']
# 1) Altura  de los hombres
altHom = hom['Height'].mean()
print('1) Altura promedio hombres:', round(altHom, 2), ':3')

# 2) Altura de las mujeres
altMuj = muj['Height'].mean()
print('2) Altura promedio mujeres:', round(altMuj, 2), ':D')

# 3) Peso  de los hombres
pesoHom = hom['Weight'].mean()
print('3) Peso promedio hombres:', round(pesoHom, 2), ':3')

# 4) Peso  de las mujeres
pesoMuj = muj['Weight'].mean()
print('4) Peso promedio mujeres:', round(pesoMuj, 2), ':D')

# 5) Mayor estatura entre los hombres
altMax = hom['Height'].max()
print('5) Mayor estatura hombres:', altMax, ':3')

# 6) Menor estatura entre las chicas
altMin = muj['Height'].min()
print('6) Menor estatura chicas:', altMin, ':D')

# 7) Peso más ligero entre los chicos de más de 1.75 m (175 cm)
homAltos = hom[hom['Height'] > 175]
pesoMin = homAltos['Weight'].min()
print('7) Peso más ligero chicos > 1.75 m:', pesoMin, ':3')

# 8) Mayor peso entre las chicas de menos de 1.60 m (160 cm)
mujBajas = muj[muj['Height'] < 160]
pesoMax = mujBajas['Weight'].max()
print('8) Mayor peso chicas < 1.60 m:', pesoMax, ':D')

# 9) Nota promedio sin faltas (N)
sinAus = est[est['Absence'] == 'N']
notaSin = sinAus['Grade'].mean()
print('9) Nota promedio sin faltas:', round(notaSin, 2), ':3')

# 10) Nota promedio con faltas (Y)
conAus = est[est['Absence'] == 'Y']
notaCon = conAus['Grade'].mean()
print('10) Nota promedio con faltas:', round(notaCon, 2), ':D')

# 11) Altura promedio chicas grupo sanguíneo A o AB
gAAB = est[est['Bloodtype'].isin(['A', 'AB'])]
altAAB = gAAB['Height'].mean()
print('11) Altura promedio de los chicos y chicas grupo A o AB:', round(altAAB, 2), ':3')

# 12) Altura promedio de chicos con grupo A o AB
homAAB = hom[hom['Bloodtype'].isin(['A', 'AB'])]
altHomAAB = homAAB['Height'].mean()
print('12) Altura promedio hombres A o AB:', round(altHomAAB, 2), ':D')

# 13) Edad promedio con faltas y nota mayor o igual a 3
ausNota = est[(est['Absence'] == 'Y') & (est['Grade'] >= 3)]
edadProm = ausNota['Age'].mean()
print('13) Edad promedio con faltas y nota >= 3:', round(edadProm, 2), ':3')
```

```python
#PROGRAMA DOS DE STEFF
import pandas as pd
import numpy as np
cen = pd.read_csv('data_census.csv')
cen.columns = ['ProvinceCode', 'CityCode', 'Population', 'Households', 'PersonsHousehold', 'GenderRatio']
cen['Population'] = cen['Population'].str.replace(',', '').str.strip().astype(int)
cen['Households'] = cen['Households'].str.replace(',', '').str.strip().astype(int)
print('Forma de la tabla:', cen.shape, ':D')
cen.head(10)
# a) Población total con ProvinceCode 115 o 116
provs = cen[cen['ProvinceCode'].isin([115, 116])]
sumaPob = provs['Population'].sum()
print('Población total provincias 115 y 116:', sumaPob, ':D/')
# b) Población promedio donde hay más hombres (GenderRatio > 1)
masHom = cen[cen['GenderRatio'] > 1]
pobProm = masHom['Population'].mean()
print('Población promedio con más hombres:', round(pobProm, 2), ':D')
# c) Más hombres y menos de 2 personas por hogar
masHomPoco = cen[(cen['GenderRatio'] > 1) & (cen['PersonsHousehold'] < 2)]
print('Estos son los lugares :3')
masHomPoco
# d) Ordeno por Households de menor a mayor y muestro 10
ordenado = cen.sort_values('Households', ascending=True)
print('Los 10 con menos hogares :D')
ordenado.head(10)
```

```python
#PROGRAMA 3 DE STEFF
import pandas as pd
import numpy as np
ven = pd.read_csv('data_sales.csv')
ven = ven.rename(columns={'Unit Price': 'UnitPrice'})
print('Forma de la tabla:', ven.shape, ':D')
ven.head(5)
# 1) Nueva columna Amount = UnitPrice x Units
ven['Amount'] = ven['UnitPrice'] * ven['Units']
print('Ya agregué Amount :3')
ven.head(5)
# 2) Precio unitario promedio por región con groupby()
grp = ven.groupby('Region')
precioReg = grp['UnitPrice'].mean()
print('Precio promedio por región :D')
print(precioReg)
# 3) Precio unitario promedio por región con pivot_table()
tabPiv = pd.pivot_table(ven, index='Region', values='UnitPrice', aggfunc='mean')
print('Con pivot_table :3')
tabPiv
# 4) Precio unitario promedio por región en una línea con groupby()
print('Una sola línea :D')
ven.groupby('Region')['UnitPrice'].mean()
# 5) Precio unitario promedio y unidades por región en una línea (pivot_table)
print('Precio y unidades :3')
pd.pivot_table(ven, index='Region', values=['UnitPrice', 'Units'], aggfunc='mean')
# 6) Total de unidades por región y artículo (faltantes = 0)
print('Unidades por región y artículo :D')
pd.pivot_table(ven, index='Region', columns='Item', values='Units', aggfunc='sum', fill_value=0)
# 7) Importe total por región y artículo (faltantes = 0)
print('Importe por región y artículo :3')
pd.pivot_table(ven, index='Region', columns='Item', values='Amount', aggfunc='sum', fill_value=0)
```

```python
#PROGRAMA 4 DE STEFF
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
caf = pd.read_csv('data_coffeeshop.csv', na_values=[' '])
caf.columns = ['yearOfStart', 'currentState', 'sizeOfSite']
print('Forma de la tabla:', caf.shape, ':D')
caf.head(5)
# Tabla de frecuencias ordenada por año
frec = caf['yearOfStart'].value_counts()
frec = frec.sort_index()
print('Cafeterías abiertas por año :3')
frec
# Gráfica de línea de 1997 a 2014
plt.plot(frec.index, frec.values)
plt.xlim(1997, 2014)
plt.xlabel('Año de inicio')
plt.ylabel('Número de cafeterías')
plt.title('Cafeterías abiertas por año :D')
plt.show()
# Separo abiertas (In) y cerradas (Out)
abi = caf[caf['currentState'] == 'In']
cer = caf[caf['currentState'] == 'Out']

# Frecuencia por año de cada grupo, ordenada por año
frecIn = abi['yearOfStart'].value_counts().sort_index()
frecOut = cer['yearOfStart'].value_counts().sort_index()
print('Ya separé las abiertas y cerradas :3')
# Dos líneas en la misma gráfica con figure
fig = plt.figure(figsize=(9, 5))
plt.plot(frecIn.index, frecIn.values, label='In (abiertas)')
plt.plot(frecOut.index, frecOut.values, label='Out (cerradas)')
plt.xlim(1997, 2014)
plt.xlabel('Año de inicio')
plt.ylabel('Número de cafeterías')
plt.title('Tendencia por estado del negocio :D')
plt.legend()
plt.show()
```
